```python
from booster_python_client.lib import BoosterLowLevelController
import booster_robotics_sdk_python as br
import booster_python_client.actions as actions
import booster_python_client.joints as joint_lists
```

```python
# Initialize Booster Client
booster = BoosterLowLevelController()
booster.init()
```

```python
# Set Mode
booster.set_mode(br.RobotMode.kPrepare)
```

```python
booster.enable_arm_usage()
```

```python
booster.send_command(actions.RIGHT_PUNCH, speed="slow", time_gap_s=0.5)
```

```python
booster.send_command(actions.LEFT_PUNCH, speed="slow", time_gap_s=0.5)
```

```python
booster.read_latest_low_state(joint_lists.LEFT_RIGHT_ARM_TORSO_JOINTS_INDICES)
```
